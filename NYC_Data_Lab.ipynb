```python
import requests
import zipfile
import io
import pandas as pd
# Download the ZIP file for January 2023
url = 'https://s3.amazonaws.com/tripdata/202401-citibike-tripdata.zip'
response = requests.get(url)

# Extract the CSV from the ZIP
with zipfile.ZipFile(io.BytesIO(response.content)) as z:
    # List files in the ZIP to find the CSV
    csv_file = [f for f in z.namelist() if f.endswith('.csv')][0]
    with z.open(csv_file) as f:
        df = pd.read_csv(f)

# Display the first few rows
df.head()
df.info()
```

```python
# Convert timestamps to datetime
df['started_at'] = pd.to_datetime(df['started_at'])
df['ended_at'] = pd.to_datetime(df['ended_at'])

# Calculate trip duration in minutes
df['trip_duration'] = (df['ended_at'] - df['started_at']).dt.total_seconds() / 60

# Remove trips with unrealistic durations (e.g., < 1 min or > 24 hours)
df = df[(df['trip_duration'] >= 1) & (df['trip_duration'] <= 1440)]

# Handle missing station names by filling with 'Unknown'
df['start_station_name'].fillna('Unknown', inplace=True)
df['end_station_name'].fillna('Unknown', inplace=True)

# Drop rows with missing coordinates
df = df.dropna(subset=['start_lat', 'start_lng', 'end_lat', 'end_lng'])

# Display cleaned data summary
df.info()
```

```python
# Extract hour from start time
df['hour'] = df['started_at'].dt.hour

# Aggregate rides by hour
hourly_rides = df.groupby('hour').size().reset_index(name='ride_count')

# Aggregate rides by start station
station_rides = df.groupby('start_station_name').agg({
    'start_lat': 'first',
    'start_lng': 'first',
    'ride_id': 'count'
}).reset_index().rename(columns={'ride_id': 'ride_count'})

# Top 10 stations by ride count
top_stations = station_rides.nlargest(10, 'ride_count')

# Aggregate rides by day of week
df['day_of_week'] = df['started_at'].dt.day_name()
day_rides = df.groupby('day_of_week').size().reset_index(name='ride_count')

# Aggregate by member type
member_rides = df.groupby('member_casual').size().reset_index(name='ride_count')

# Display sample aggregated data
hourly_rides.head()

most_popular = station_rides.loc[station_rides["ride_count"].idxmax()]

print(f"Station with the most rides: {most_popular['start_station_name']}")
print(f"Number of rides: {most_popular['ride_count']}")
```

```python

import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
# Create a subplot layout (2 rows, 3 columns)
fig = make_subplots(
    rows=2, cols=3,
    specs=[[{"type": "xy"}, {"type": "xy"}, {"type": "xy"}],
           [{"type": "domain"}, {"type": "xy"}, {"type": "scattergeo"}]],
    subplot_titles=("Rides by Hour", "Top 10 Stations", "Trip Duration Distribution",
                    "Rides by Member Type", "Rides by Day", "Station Map")
)

# 1. Line Chart: Rides by Hour
fig.add_trace(
    go.Scatter(x=hourly_rides['hour'], y=hourly_rides['ride_count'], mode='lines+markers',
               name='Rides'),
    row=1, col=1
)

# 2. Bar Chart: Top 10 Stations
fig.add_trace(
    go.Bar(x=top_stations['start_station_name'], y=top_stations['ride_count'],
           name='Stations'),
    row=1, col=2
)

# 3. Box Plot: Trip Duration Distribution
fig.add_trace(
    go.Box(y=df['trip_duration'], name='Duration', boxpoints='outliers'),
    row=1, col=3
)

# 4. Pie Chart: Rides by Member Type
fig.add_trace(
    go.Pie(labels=member_rides['member_casual'], values=member_rides['ride_count'],
           name='Member Type'),
    row=2, col=1
)

# 5. Bar Chart: Rides by Day of Week
fig.add_trace(
    go.Bar(x=day_rides['day_of_week'], y=day_rides['ride_count'], name='Day'),
    row=2, col=2
)

# 6. Geoplot: Station Map
fig.add_trace(
    go.Scattergeo(
        lon=station_rides['start_lng'],
        lat=station_rides['start_lat'],
        text=station_rides['start_station_name'],
        mode='markers',
        marker=dict(
            size=station_rides['ride_count'] / station_rides['ride_count'].max() * 20,
            color=station_rides['ride_count'],
            colorscale='Viridis',
            showscale=True
        ),
        name='Stations'
    ),
    row=2, col=3
)

# Update geoplot layout
fig.update_geos(
    scope='usa',
    #projection_type='mercator',
    center=dict(lat=40.7128, lon=-74.0060),  # NYC coordinates
    lataxis_range=[40.5, 41.0],
    lonaxis_range=[-74.3, -73.7]
)

# Update overall layout
fig.update_layout(
    title_text='NYC Citi Bike Trip Data Dashboard',
    showlegend=False,
    height=800
)

# Show the dashboard
fig.show()
```

```python
# Calculate average trip duration for each hour
avg_duration_by_hour = (
    df.groupby("hour")["trip_duration"]
      .mean()
      .reset_index(name="average_trip_duration")
)

# Create the line chart
fig1 = px.line(
    avg_duration_by_hour,
    x="hour",
    y="average_trip_duration",
    markers=True,
    title="Average Trip Duration by Hour of Day",
    labels={
        "hour": "Hour of Day",
        "average_trip_duration": "Average Trip Duration (Minutes)"
    }
)

fig1.update_xaxes(dtick=1)
fig1.show()
```

```python
# Identify the five stations with the most rides
top_5_station_names = (
    df.groupby("start_station_name")
      .size()
      .nlargest(5)
      .index
)

# Filter the original DataFrame
top_5_rides = df[
    df["start_station_name"].isin(top_5_station_names)
]

# Count rides by station and member type
station_member_counts = (
    top_5_rides
    .groupby(["start_station_name", "member_casual"])
    .size()
    .reset_index(name="ride_count")
)

# Create a grouped bar chart
fig2 = px.bar(
    station_member_counts,
    x="start_station_name",
    y="ride_count",
    color="member_casual",
    barmode="group",
    title="Ride Counts at the Top Five Stations by Member Type",
    labels={
        "start_station_name": "Starting Station",
        "ride_count": "Number of Rides",
        "member_casual": "Member Type"
    }
)

fig2.update_layout(xaxis_tickangle=-35)
fig2.show()
```

```python
import numpy as np

def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Earth radius in km
    dlat = np.radians(lat2 - lat1)
    dlon = np.radians(lon2 - lon1)
    a = np.sin(dlat/2)**2 + np.cos(np.radians(lat1)) * np.cos(np.radians(lat2)) * np.sin(dlon/2)**2
    c = 2 * np.arctan2(np.sqrt(a), np.sqrt(1-a))
    return R * c
```
